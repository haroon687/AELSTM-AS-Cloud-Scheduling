```python
SQL = r'''WITH usage_rows AS (
 SELECT machine_id, DIV(start_time, 300000000) AS bucket_index,
        CAST(average_usage.cpus AS FLOAT64) AS cpu_usage
 FROM `google.com:google-cluster-data.clusterdata_2019_a.instance_usage`
 WHERE start_time >= 0 AND end_time <= 86400000000
   AND machine_id IS NOT NULL AND average_usage.cpus IS NOT NULL
   AND average_usage.cpus >= 0
   AND (alloc_collection_id IS NULL OR alloc_collection_id = 0)
   AND end_time - start_time = 300000000
   AND MOD(start_time, 300000000) = 0
), usage_by_machine AS (
 SELECT machine_id, bucket_index, SUM(cpu_usage) AS cpu_demand,
        COUNT(*) AS contributing_instances
 FROM usage_rows GROUP BY machine_id, bucket_index
), capacity_events AS (
 SELECT machine_id, time, type, CAST(capacity.cpus AS FLOAT64) AS cpu_capacity
 FROM `google.com:google-cluster-data.clusterdata_2019_a.machine_events`
 WHERE time <= 86400000000 AND machine_id IS NOT NULL
), joined AS (
 SELECT u.machine_id, u.bucket_index, u.cpu_demand, u.contributing_instances,
        e.type AS machine_event_type, e.cpu_capacity,
        ROW_NUMBER() OVER (
          PARTITION BY u.machine_id, u.bucket_index ORDER BY e.time DESC
        ) AS event_rank
 FROM usage_by_machine u
 LEFT JOIN capacity_events e
   ON e.machine_id = u.machine_id
  AND e.time <= u.bucket_index * 300000000
)
SELECT machine_id, bucket_index, cpu_demand, cpu_capacity,
       SAFE_DIVIDE(cpu_demand, cpu_capacity) AS cpu_utilization_raw,
       contributing_instances
FROM joined
WHERE event_rank = 1 AND machine_event_type IN (1, 3)
  AND cpu_capacity > 0
ORDER BY machine_id, bucket_index'''
dry_config = bigquery.QueryJobConfig(dry_run=True, use_query_cache=False)
dry_job = client.query(SQL, job_config=dry_config)
estimated_gib = dry_job.total_bytes_processed / (1024**3)
print(f'Estimated processing: {estimated_gib:.3f} GiB')
print(f'Estimated bytes: {dry_job.total_bytes_processed:,}')
print('No extraction has been run yet.')
```

```python
import os
import zipfile
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from google.colab import files
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from tensorflow.keras import Model
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

SEED = 42
SEQ_LEN = 20
BATCH_SIZE = 64
EPOCHS = 100
LEARNING_RATE = 0.001

tf.keras.utils.set_random_seed(SEED)
np.random.seed(SEED)

OUTPUT_DIR = "figure2_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("TensorFlow:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices("GPU"))
```

```python
df["bucket_index"] = pd.to_numeric(df["bucket_index"], errors="coerce")
df["cpu_demand"] = pd.to_numeric(df["cpu_demand"], errors="coerce")
df["cpu_capacity"] = pd.to_numeric(df["cpu_capacity"], errors="coerce")
df["cpu_utilization_raw"] = pd.to_numeric(
    df["cpu_utilization_raw"], errors="coerce"
)

df = df.dropna(subset=[
    "machine_id", "bucket_index", "cpu_demand",
    "cpu_capacity", "cpu_utilization_raw"
]).copy()

df = df[df["cpu_capacity"] > 0].copy()

# Convert fractional utilization to percentage points when appropriate.
q99 = df["cpu_utilization_raw"].quantile(0.99)
if q99 <= 1.5:
    df["cpu_utilization_pct"] = df["cpu_utilization_raw"] * 100.0
    print("Detected fractional CPU utilization; converted to percentage.")
else:
    df["cpu_utilization_pct"] = df["cpu_utilization_raw"]
    print("Using CPU utilization values as percentage points.")

df["bucket_index"] = df["bucket_index"].astype(int)
df = (
    df.sort_values(["machine_id", "bucket_index"])
      .drop_duplicates(["machine_id", "bucket_index"], keep="last")
      .reset_index(drop=True)
)

if (df["cpu_utilization_pct"] > 100).any():
    n_over = int((df["cpu_utilization_pct"] > 100).sum())
    warnings.warn(
        f"{n_over} rows exceed 100% CPU utilization. They are retained "
        "for inspection; check the trace units and aggregation."
    )

print("Clean rows:", len(df))
print("Machines:", df["machine_id"].nunique())
print("Bucket range:", df["bucket_index"].min(), "to", df["bucket_index"].max())
display(df[["cpu_demand", "cpu_capacity", "cpu_utilization_pct"]].describe())

plt.figure(figsize=(8, 4))
plt.hist(df["cpu_utilization_pct"].dropna(), bins=50)
plt.xlabel("CPU utilization (%)")
plt.ylabel("Number of machine-buckets")
plt.title("Google Cluster Trace: CPU utilization distribution")
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/Fig2a_CPU_Distribution.png", dpi=300)
plt.show()
```

```python
X_list, y_list = [], []
machine_list, target_bucket_list = [], []

for machine_id, g in df.groupby("machine_id", sort=False):
    g = g.sort_values("bucket_index").reset_index(drop=True)
    values = g["cpu_utilization_pct"].to_numpy(dtype=np.float32)
    buckets = g["bucket_index"].to_numpy(dtype=np.int64)

    for end_idx in range(SEQ_LEN, len(g)):
        window_buckets = buckets[end_idx - SEQ_LEN : end_idx + 1]
        # 20 input buckets plus the immediately following target bucket.
        if not np.all(np.diff(window_buckets) == 1):
            continue

        X_list.append(values[end_idx - SEQ_LEN : end_idx])
        y_list.append(values[end_idx])
        machine_list.append(machine_id)
        target_bucket_list.append(buckets[end_idx])

if not X_list:
    raise ValueError(
        "No contiguous 20-step sequences were found. "
        "Check bucket_index continuity and the selected extraction interval."
    )

X_raw = np.asarray(X_list, dtype=np.float32)[..., np.newaxis]
y_raw = np.asarray(y_list, dtype=np.float32)
machine_ids = np.asarray(machine_list)
target_buckets = np.asarray(target_bucket_list, dtype=np.int64)

print("Sequence tensor:", X_raw.shape)
print("Target vector:", y_raw.shape)
print("Machines represented:", len(np.unique(machine_ids)))
```

```python
bucket_min = int(df["bucket_index"].min())
bucket_max = int(df["bucket_index"].max())
train_cutoff = bucket_min + int(0.70 * (bucket_max - bucket_min))
val_cutoff = bucket_min + int(0.85 * (bucket_max - bucket_min))

train_mask = target_buckets <= train_cutoff
val_mask = (target_buckets > train_cutoff) & (target_buckets <= val_cutoff)
test_mask = target_buckets > val_cutoff

if train_mask.sum() == 0 or val_mask.sum() == 0 or test_mask.sum() == 0:
    raise ValueError(
        "A chronological split is empty. The trace may be too short; "
        "use a longer extraction interval or adjust the split proportions."
    )

# Fit the scaler only on observations in the training time period.
train_observations = df.loc[
    df["bucket_index"] <= train_cutoff, "cpu_utilization_pct"
].to_numpy().reshape(-1, 1)

scaler = MinMaxScaler()
scaler.fit(train_observations)

X_scaled = scaler.transform(X_raw.reshape(-1, 1)).reshape(X_raw.shape)
y_scaled = scaler.transform(y_raw.reshape(-1, 1)).reshape(-1)

X_train, y_train = X_scaled[train_mask], y_scaled[train_mask]
X_val, y_val = X_scaled[val_mask], y_scaled[val_mask]
X_test, y_test = X_scaled[test_mask], y_scaled[test_mask]

print("Train samples:", len(X_train))
print("Validation samples:", len(X_val))
print("Test samples:", len(X_test))
print("Training cutoff bucket:", train_cutoff)
print("Validation cutoff bucket:", val_cutoff)
```

```python
class TemporalAttention(tf.keras.layers.Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.score_layer = Dense(1, activation="tanh")

    def call(self, hidden_states):
        # hidden_states shape: (batch, time_steps, hidden_units)
        scores = self.score_layer(hidden_states)        # (batch, time_steps, 1)
        weights = tf.nn.softmax(scores, axis=1)         # temporal attention
        context = tf.reduce_sum(
            hidden_states * weights, axis=1
        )                                               # (batch, hidden_units)
        return context, tf.squeeze(weights, axis=-1)


inputs = Input(shape=(SEQ_LEN, 1), name="historical_cpu_sequence")
h = LSTM(128, return_sequences=True, name="lstm_layer_1")(inputs)
h = LSTM(128, return_sequences=True, name="lstm_layer_2")(h)

context_vector, attention_weights = TemporalAttention(
    name="temporal_attention"
)(h)

dense_features = Dense(64, activation="relu", name="dense_prediction_features")(
    context_vector
)
dense_features = Dropout(0.20, name="dropout")(dense_features)
forecast = Dense(1, activation="linear", name="next_step_cpu_prediction")(
    dense_features
)

model = Model(inputs=inputs, outputs=forecast, name="AELSTM_Google_Cluster")
attention_model = Model(inputs=inputs, outputs=attention_weights)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=LEARNING_RATE),
    loss="mse",
    metrics=["mae"]
)

model.summary()
```

```python
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[early_stop],
    verbose=1
)

plt.figure(figsize=(8, 4))
plt.plot(history.history["loss"], label="Training loss")
plt.plot(history.history["val_loss"], label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("Mean squared error")
plt.title("AELSTM training and validation loss")
plt.legend()
plt.tight_layout()
plt.savefig(f"{OUTPUT_DIR}/Fig2b_Training_Validation_Loss.png", dpi=300)
plt.show()
```

```python
# Add actual incoming task CPU demands here, in the same units as cpu_demand/cpu_capacity.
# Leave empty until you have task-level requirements.
TASK_CPU_REQUIREMENTS = []

# Configurable demonstration settings; calibrate/validate before manuscript reporting.
U_TARGET = 70.0
U_MAX = 90.0
ALPHA = 0.40
BETA = 0.40
GAMMA = 0.20

def prediction_driven_schedule(vm_table, task_cpu_requirements,
                               u_target=U_TARGET, u_max=U_MAX,
                               alpha=ALPHA, beta=BETA, gamma=GAMMA):
    if vm_table.empty:
        raise ValueError("No valid VM forecasts are available.")
    if not task_cpu_requirements:
        print("Scheduler skipped: TASK_CPU_REQUIREMENTS is empty.")
        return pd.DataFrame()

    state = vm_table.copy().reset_index(drop=True)
    if (state["cpu_capacity"] <= 0).any():
        raise ValueError("All VM capacities must be positive.")

    decisions = []

    for task_no, task_demand in enumerate(task_cpu_requirements, start=1):
        task_demand = float(task_demand)
        if task_demand < 0:
            raise ValueError("Task CPU requirements cannot be negative.")

        predicted = state["predicted_cpu_utilization_pct"].to_numpy(dtype=float)
        capacities = state["cpu_capacity"].to_numpy(dtype=float)

        feasible = []
        for j in range(len(state)):
            projected_j = predicted[j] + (task_demand / capacities[j]) * 100.0
            if projected_j <= u_max:
                projected_vector = predicted.copy()
                projected_vector[j] = projected_j

                target_cost = abs(projected_j - u_target) / 100.0
                imbalance_cost = np.std(projected_vector) / 100.0
                pressure_cost = projected_j / u_max

                total_cost = (
                    alpha * target_cost
                    + beta * imbalance_cost
                    + gamma * pressure_cost
                )

                feasible.append((total_cost, j, projected_j,
                                 target_cost, imbalance_cost, pressure_cost))

        if not feasible:
            decisions.append({
                "task_id": task_no,
                "task_cpu_requirement": task_demand,
                "selected_machine_id": None,
                "decision": "DEFERRED_NO_FEASIBLE_VM",
                "projected_utilization_pct": np.nan,
                "decision_cost": np.nan
            })
            continue

        best = min(feasible, key=lambda z: z[0])
        total_cost, j, projected_j, t_cost, b_cost, p_cost = best
        selected_machine = state.loc[j, "machine_id"]

        decisions.append({
            "task_id": task_no,
            "task_cpu_requirement": task_demand,
            "selected_machine_id": selected_machine,
            "decision": "ALLOCATED",
            "projected_utilization_pct": projected_j,
            "decision_cost": total_cost,
            "target_deviation_cost": t_cost,
            "load_imbalance_cost": b_cost,
            "capacity_pressure_cost": p_cost
        })

        # Update the predicted utilization state after this allocation.
        state.loc[j, "predicted_cpu_utilization_pct"] = projected_j

    return pd.DataFrame(decisions)


if TASK_CPU_REQUIREMENTS and not vm_state.empty:
    schedule_results = prediction_driven_schedule(vm_state, TASK_CPU_REQUIREMENTS)
    display(schedule_results)
    schedule_results.to_csv(
        f"{OUTPUT_DIR}/Adaptive_Scheduling_Demonstration.csv", index=False
    )
else:
    schedule_results = pd.DataFrame()
    print(
        "To activate this section, populate TASK_CPU_REQUIREMENTS with "
        "real task CPU demands in the same units as cpu_capacity."
    )
```

```python
def confidence_interval(values, confidence=0.95):
    values = np.asarray(values, dtype=float)
    n = len(values)

    mean = np.mean(values)
    sem = stats.sem(values)
    margin = stats.t.ppf(
        (1 + confidence) / 2, n - 1
    ) * sem

    return mean - margin, mean + margin


summary_rows = []

for metric in METRICS:
    for method in METHODS:

        values = results_df.loc[
            results_df["Method"] == method, metric
        ].to_numpy()

        ci_low, ci_high = confidence_interval(values)

        summary_rows.append({
            "Metric": metric,
            "Method": method,
            "Mean": np.mean(values),
            "Std": np.std(values, ddof=1),
            "CI_95_Lower": ci_low,
            "CI_95_Upper": ci_high
        })

summary_df = pd.DataFrame(summary_rows)

display(summary_df.round(4))

summary_df.to_csv(
    "AELSTM_Mean_Std_95CI.csv",
    index=False
)
```

```python
comparison_rows = []

for metric in METRICS:

    reference = (
        results_df[results_df["Method"] == "AELSTM-AS"]
        .set_index("Seed")[metric]
        .sort_index()
    )

    raw_p_values = []
    temporary_rows = []

    for method in METHODS:

        if method == "AELSTM-AS":
            continue

        comparison = (
            results_df[results_df["Method"] == method]
            .set_index("Seed")[metric]
            .sort_index()
        )

        # Paired differences: baseline minus AELSTM-AS
        differences = comparison.values - reference.values

        # Paired t-test
        t_stat, p_t = stats.ttest_rel(
            comparison.values,
            reference.values
        )

        # Wilcoxon signed-rank test
        if np.allclose(differences, 0):
            w_stat, p_w = 0.0, 1.0
        else:
            w_stat, p_w = stats.wilcoxon(
                comparison.values,
                reference.values,
                alternative="two-sided",
                zero_method="wilcox"
            )

        # Paired Cohen's dz
        sd_diff = np.std(differences, ddof=1)

        if sd_diff == 0:
            cohen_dz = 0.0
        else:
            cohen_dz = np.mean(differences) / sd_diff

        raw_p_values.append(p_t)

        temporary_rows.append({
            "Metric": metric,
            "Comparison": method + " vs AELSTM-AS",
            "Paired_t_statistic": t_stat,
            "Paired_t_p_raw": p_t,
            "Wilcoxon_statistic": w_stat,
            "Wilcoxon_p": p_w,
            "Paired_Cohens_dz": cohen_dz
        })

    # Holm correction across the five baseline comparisons
    adjusted = multipletests(
        raw_p_values,
        method="holm"
    )[1]

    for row, adjusted_p in zip(temporary_rows, adjusted):
        row["Paired_t_p_Holm"] = adjusted_p
        comparison_rows.append(row)

paired_df = pd.DataFrame(comparison_rows)

display(paired_df.round(5))

paired_df.to_csv(
    "AELSTM_Paired_Statistical_Tests.csv",
    index=False
)
```

```python
import requests
import gzip
import os
import pyarrow.parquet as pq

# STEP 1: Download the verified public Parquet file

bucket = "clusterdata_2019_a"
file_name = "instance_usage-0000000000.parquet.gz"

list_url = f"https://storage.googleapis.com/storage/v1/b/{bucket}/o"

params = {
    "prefix": "instance_usage-0000000000",
    "maxResults": 20
}

response = requests.get(list_url, params=params, timeout=60)
response.raise_for_status()

items = response.json().get("items", [])

matching_files = [
    item for item in items
    if item["name"] == file_name
]

if not matching_files:
    raise Exception("Required Parquet file not found.")

selected = matching_files[0]

print("Verified file:", selected["name"])
print("Expected size (MB):",
      round(int(selected["size"]) / (1024**2), 2))

# STEP 2: Download

download_url = selected["mediaLink"]
file_path = "/content/" + file_name

download = requests.get(
    download_url,
    stream=True,
    timeout=180
)

download.raise_for_status()

with open(file_path, "wb") as f:
    for chunk in download.iter_content(1024 * 1024):
        if chunk:
            f.write(chunk)

print("\nDownload completed.")
print("Actual size (MB):",
      round(os.path.getsize(file_path) / (1024**2), 2))

# STEP 3: Decompress

parquet_path = "/content/instance_usage_sample.parquet"

with gzip.open(file_path, "rb") as source:
    with open(parquet_path, "wb") as target:
        while True:
            chunk = source.read(1024 * 1024)
            if not chunk:
                break
            target.write(chunk)

print("\nDecompression successful.")

# STEP 4: Inspect Parquet metadata

parquet_file = pq.ParquetFile(parquet_path)

print("\n========== DATASET INFORMATION ==========")
print("Number of rows:", parquet_file.metadata.num_rows)
print("Number of columns:", parquet_file.metadata.num_columns)

print("\n========== DATASET SCHEMA ==========")
print(parquet_file.schema)
```

```python
# STEP 3: Extract CPU usage from Google Cluster Trace

import pyarrow.parquet as pq
import pandas as pd

# Use the file discovered in the previous successful cell
parquet_file = pq.ParquetFile(input_file)

cpu_records = []

print("Starting CPU extraction...")

# Process the first 2 row groups initially
for group_no in range(min(2, parquet_file.num_row_groups)):

    print("Processing row group:", group_no)

    table = parquet_file.read_row_group(group_no)

    df = table.to_pandas()

    # Extract CPU usage from nested average_usage
    for _, row in df.iterrows():

        usage = row["average_usage"]

        if isinstance(usage, dict):

            cpu_value = usage.get("cpus", None)

            if cpu_value is not None:

                cpu_records.append({
                    "start_time": row["start_time"],
                    "end_time": row["end_time"],
                    "machine_id": row["machine_id"],
                    "instance_index": row["instance_index"],
                    "cpu_usage": cpu_value
                })

    del df, table

# Create extracted CPU dataset
df_cpu = pd.DataFrame(cpu_records)

print("\nCPU EXTRACTION SUMMARY")
print("Total extracted records:", len(df_cpu))

display(df_cpu.head(10))
```

```python
# STEP 4: Validate extracted CPU dataset

import pandas as pd
import numpy as np

print("DATASET VALIDATION")
print("=" * 45)

print("Total records:", len(df_cpu))

print("\nMissing values:")
print(df_cpu.isnull().sum())

print("\nDuplicate records:")
print(df_cpu.duplicated().sum())

print("\nCPU usage statistics:")
print(df_cpu["cpu_usage"].describe())

print("\nNegative CPU values:",
      (df_cpu["cpu_usage"] < 0).sum())

print("\nTime interval statistics:")
df_cpu["interval"] = (
    df_cpu["end_time"] - df_cpu["start_time"]
)

print(df_cpu["interval"].describe())

# Save extracted data
output_path = "/content/Google_Cluster_CPU_Extracted.csv"

df_cpu.to_csv(output_path, index=False)

print("\nCSV SAVED SUCCESSFULLY")
print("File:", output_path)
```

```python
# STEP 5: Clean and prepare CPU time-series records

import pandas as pd
import numpy as np

# Create a working copy
df_clean = df_cpu.copy()

# Remove exact duplicate records
before = len(df_clean)

df_clean = df_clean.drop_duplicates(
    subset=[
        "start_time",
        "end_time",
        "machine_id",
        "instance_index",
        "cpu_usage"
    ]
)

after = len(df_clean)

# Sort records chronologically
df_clean = df_clean.sort_values(
    by=["start_time", "machine_id"]
).reset_index(drop=True)

# Convert microseconds into seconds
df_clean["time_seconds"] = (
    df_clean["start_time"] / 1_000_000
)

df_clean["duration_seconds"] = (
    df_clean["end_time"] - df_clean["start_time"]
) / 1_000_000

print("DATA CLEANING SUMMARY")
print("-----------------------------")
print("Records before cleaning:", before)
print("Records after cleaning:", after)
print("Duplicate records removed:", before - after)

print("\nTIME CONVERSION")
print(df_clean[
    ["start_time", "time_seconds", "duration_seconds"]
].head(10))

print("\nCLEANED DATASET")
display(df_clean.head(10))

# Save cleaned data
clean_file = "/content/Google_Cluster_CPU_Cleaned.csv"

df_clean.to_csv(clean_file, index=False)

print("\nCleaned dataset saved:", clean_file)
```

```python
# STEP 6: Aggregate CPU workload by time interval

import pandas as pd
import numpy as np

# Aggregate CPU demand at each start time
df_time = (
    df_clean.groupby("start_time")
    .agg(
        total_cpu_demand=("cpu_usage", "sum"),
        mean_cpu_demand=("cpu_usage", "mean"),
        max_cpu_demand=("cpu_usage", "max"),
        active_instances=("instance_index", "count"),
        active_machines=("machine_id", "nunique")
    )
    .reset_index()
)

# Convert timestamp into seconds
df_time["time_seconds"] = (
    df_time["start_time"] / 1_000_000
)

# Sort chronologically
df_time = df_time.sort_values(
    "start_time"
).reset_index(drop=True)

print("TIME-SERIES AGGREGATION SUMMARY")
print("=" * 45)

print("Total time steps:", len(df_time))
print("Unique timestamps:", df_time["start_time"].nunique())

print("\nFirst 10 aggregated records:")
display(df_time.head(10))

print("\nCPU demand statistics:")
display(df_time[
    ["total_cpu_demand", "mean_cpu_demand",
     "max_cpu_demand", "active_instances", "active_machines"]
].describe())

# Save aggregated workload
df_time.to_csv(
    "/content/Google_Cluster_CPU_TimeSeries.csv",
    index=False
)

print("\nTime-series CSV saved successfully.")
```

```python
# STEP 7: Analyze time-series sampling intervals

import numpy as np
import pandas as pd

# Calculate time difference between consecutive observations
df_time = df_time.sort_values("time_seconds").reset_index(drop=True)

df_time["time_gap_seconds"] = (
    df_time["time_seconds"].diff()
)

# Remove the first NaN value
gaps = df_time["time_gap_seconds"].dropna()

print("TIME INTERVAL ANALYSIS")
print("=" * 45)

print("Total observations:", len(df_time))
print("Minimum time gap:", gaps.min(), "seconds")
print("Maximum time gap:", gaps.max(), "seconds")
print("Mean time gap:", gaps.mean(), "seconds")
print("Median time gap:", gaps.median(), "seconds")

print("\nTime gap distribution:")
print(
    gaps.describe(
        percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
    )
)

print("\nMost frequent time gaps:")
print(gaps.value_counts().head(15))

# Check whether the observations are regularly spaced
print(
    "\nAre all time gaps equal?",
    gaps.nunique() == 1
)
```

```python
# STEP 8: Convert irregular observations into 60-second intervals

import pandas as pd
import numpy as np

# Work on a copy of the aggregated dataset
df_resample = df_time.copy()

# Sort observations by timestamp
df_resample = df_resample.sort_values(
    "time_seconds"
).reset_index(drop=True)

# Convert numeric seconds into a time index
df_resample["time_index"] = pd.to_timedelta(
    df_resample["time_seconds"],
    unit="s"
)

df_resample = df_resample.set_index("time_index")

# Resample into fixed 60-second intervals
df_60s = df_resample.resample("60s").agg({
    "total_cpu_demand": "mean",
    "mean_cpu_demand": "mean",
    "max_cpu_demand": "max",
    "active_instances": "mean",
    "active_machines": "mean"
})

# Count observations in each interval
df_60s["observation_count"] = (
    df_resample["total_cpu_demand"]
    .resample("60s")
    .count()
)

# Restore timestamp in seconds
df_60s["time_seconds"] = (
    df_60s.index.total_seconds()
)

print("60-SECOND RESAMPLING SUMMARY")
print("=" * 45)

print("Original observations:", len(df_time))
print("Resampled time intervals:", len(df_60s))
print(
    "Intervals containing observations:",
    (df_60s["observation_count"] > 0).sum()
)
print(
    "Empty intervals:",
    (df_60s["observation_count"] == 0).sum()
)

print("\nFirst 15 resampled observations:")
display(df_60s.head(15))

print("\nMissing values:")
print(df_60s.isnull().sum())
```

```python

# STEP 9A: Missing Interval Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Work on a copy of the resampled dataset
df_missing = df_60s.copy()

# Identify intervals with and without observations
df_missing["is_missing"] = (
    df_missing["observation_count"] == 0
)

total_intervals = len(df_missing)
missing_intervals = int(df_missing["is_missing"].sum())
valid_intervals = total_intervals - missing_intervals

missing_percentage = (
    missing_intervals / total_intervals
) * 100

print("=" * 55)
print("MISSING INTERVAL ANALYSIS")
print("=" * 55)

print("Total intervals:", total_intervals)
print("Valid intervals:", valid_intervals)
print("Missing intervals:", missing_intervals)
print("Missing percentage:", round(missing_percentage, 2), "%")

# Check missing intervals in consecutive groups
missing_flag = df_missing["is_missing"].astype(int)

group_id = (
    missing_flag.ne(missing_flag.shift())
    .cumsum()
)

missing_runs = (
    df_missing[missing_flag == 1]
    .groupby(group_id)
    .size()
)

print("\nConsecutive missing interval analysis")
print("--------------------------------------")
print("Number of missing groups:", len(missing_runs))

if len(missing_runs) > 0:
    print("Longest missing group:", missing_runs.max(), "minutes")
    print("Average missing group:", round(missing_runs.mean(), 2), "minutes")
else:
    print("No missing intervals found.")

# Check missing values in workload features
print("\nMissing values by feature")
print("--------------------------------------")
print(df_missing.isnull().sum())

# Plot observed and missing intervals
plt.figure(figsize=(12, 4))

plt.plot(
    df_missing["time_seconds"] / 3600,
    df_missing["observation_count"],
    linewidth=0.8
)

plt.xlabel("Time (hours)")
plt.ylabel("Observations per interval")
plt.title("Distribution of Observations Across 60-Second Intervals")
plt.grid(False)
plt.tight_layout()
plt.show()

# Plot missing interval distribution
plt.figure(figsize=(6, 4))

plt.bar(
    ["Valid intervals", "Missing intervals"],
    [valid_intervals, missing_intervals]
)

plt.ylabel("Number of intervals")
plt.title("Valid vs Missing 60-Second Intervals")
plt.grid(False)
plt.tight_layout()
plt.show()
```

```python

# STEP 9B: Controlled treatment of missing intervals

import pandas as pd
import numpy as np

# Make a fresh copy
df_clean = df_60s.copy()

# Workload features used for preprocessing
feature_cols = [
    "total_cpu_demand",
    "mean_cpu_demand",
    "max_cpu_demand",
    "active_instances",
    "active_machines"
]

# Identify original missing intervals
original_missing = (
    df_clean["observation_count"] == 0
)

# Preserve missing-data information
df_clean["was_missing"] = original_missing.astype(int)

print("Original missing intervals:",
      int(original_missing.sum()))

# Time-based interpolation for internal gaps only.
# Limit = 2 means at most two consecutive missing minutes.
df_clean[feature_cols] = (
    df_clean[feature_cols]
    .interpolate(
        method="time",
        limit=2,
        limit_area="inside"
    )
)

# Count how many originally missing intervals were filled
imputed_mask = (
    original_missing &
    df_clean[feature_cols].notna().all(axis=1)
)

print("Successfully imputed intervals:",
      int(imputed_mask.sum()))

# Identify remaining missing intervals
remaining_mask = df_clean[feature_cols].isna().any(axis=1)

print("Remaining incomplete intervals:",
      int(remaining_mask.sum()))

# Exclude intervals that remain incomplete
df_clean = df_clean.loc[~remaining_mask].copy()

# Retain the original missing indicator and time information
df_clean = df_clean.sort_values("time_seconds")

# Final data-quality report
print("\nFINAL CLEANING SUMMARY")
print("=" * 45)

print("Rows before cleaning:", len(df_60s))
print("Rows after cleaning:", len(df_clean))
print("Rows removed:", len(df_60s) - len(df_clean))

print("Imputed intervals:", int(df_clean["was_missing"].sum()))

print("\nRemaining missing values:")
print(df_clean[feature_cols].isnull().sum())

print("\nCleaned dataset preview:")
display(df_clean.head(10))

# Save intermediate cleaned dataset
df_clean.to_csv(
    "/content/google_cluster_60s_cleaned.csv",
    index=True
)

print("\nSaved: google_cluster_60s_cleaned.csv")
```

```python

# STEP 10G: Reconstruct active cluster CPU capacity

import pandas as pd
import numpy as np

ev = events.sort_values(["time", "machine_id"]).copy()

active_machines = {}
total_capacity = 0.0
capacity_records = []

for row in ev.itertuples(index=False):

    mid = row.machine_id
    event_type = int(row.type)
    new_capacity = row.capacity_cpus

    if event_type == 1:  # ADD
        if pd.notna(new_capacity):
            if mid not in active_machines:
                active_machines[mid] = float(new_capacity)
                total_capacity += float(new_capacity)

    elif event_type == 2:  # REMOVE
        if mid in active_machines:
            total_capacity -= active_machines.pop(mid)

    elif event_type == 3:  # UPDATE
        if mid in active_machines and pd.notna(new_capacity):
            old_capacity = active_machines[mid]
            active_machines[mid] = float(new_capacity)
            total_capacity += float(new_capacity) - old_capacity

    capacity_records.append({
        "time": row.time,
        "active_machine_count": len(active_machines),
        "total_cpu_capacity": total_capacity
    })

df_capacity = pd.DataFrame(capacity_records)

# Remove duplicate timestamps, retaining the last event state
df_capacity = (
    df_capacity.sort_values("time")
    .drop_duplicates("time", keep="last")
    .reset_index(drop=True)
)

print("CLUSTER CAPACITY RECONSTRUCTION")
print("=" * 50)
print("Capacity snapshots:", len(df_capacity))
print("Final active machines:", len(active_machines))
print("Final total CPU capacity:", round(total_capacity, 6))

print("\nCAPACITY STATISTICS")
display(df_capacity[
    ["active_machine_count", "total_cpu_capacity"]
].describe())

print("\nFIRST 15 CAPACITY SNAPSHOTS")
display(df_capacity.head(15))

# Save reconstructed capacity history
df_capacity.to_csv(
    "/content/google_cluster_active_cpu_capacity.csv",
    index=False
)

print("\nSaved: google_cluster_active_cpu_capacity.csv")
```

```python

import pandas as pd
import numpy as np

print("WORKLOAD-SPECIFIC CPU CAPACITY RECONSTRUCTION")
print("=" * 65)

# 1. Select machines appearing in the workload
workload_ids = set(df_cpu["machine_id"].dropna().unique())

# 2. Keep only relevant machine events
events = df_machine[
    df_machine["machine_id"].isin(workload_ids)
].copy()

# 3. Convert event timestamps from microseconds to seconds
events["time_seconds"] = events["time"] / 1_000_000

# 4. Sort machine events chronologically
events = events.sort_values(
    ["time_seconds", "machine_id"]
).reset_index(drop=True)

# 5. Verify event types
print("\nMachine event types:")
print(events["type"].value_counts().sort_index())

# Google Cluster Trace event types:
# 1 = ADD, 2 = REMOVE, 3 = UPDATE

# 6. Reconstruct active capacity for each workload timestamp
# Use the workload timeline from the cleaned 60-second dataset.

timeline = np.sort(
    df_60s["time_seconds"].dropna().unique()
)

active_capacity = {}
capacity_records = []

for timestamp in timeline:

    current_events = events[
        events["time_seconds"] <= timestamp
    ]

    # Process events up to the current timestamp.
    # Replaying only newly encountered events is more efficient.
    if "last_processed" not in locals():
        last_processed = 0

    new_events = current_events.iloc[last_processed:]
    last_processed = len(current_events)

    for row in new_events.itertuples(index=False):

        machine_id = row.machine_id
        event_type = row.type
        cpu_capacity = row.capacity_cpus

        if event_type == 1:       # ADD
            active_capacity[machine_id] = cpu_capacity

        elif event_type == 2:     # REMOVE
            active_capacity.pop(machine_id, None)

        elif event_type == 3:     # UPDATE
            if machine_id in active_capacity:
                active_capacity[machine_id] = cpu_capacity

    capacity_records.append({
        "time_seconds": timestamp,
        "active_workload_machines": len(active_capacity),
        "workload_cpu_capacity": sum(active_capacity.values())
    })

df_workload_capacity = pd.DataFrame(capacity_records)

print("\nCAPACITY RECONSTRUCTION SUMMARY")
print("Rows:", len(df_workload_capacity))
print(df_workload_capacity.describe())

display(df_workload_capacity.head(10))
```

```python

# ==========================================================
# AELSTM-AS: WORKLOAD FORECASTING EXPERIMENT
# Google Cluster Trace - 60-second aggregated dataset
# ==========================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, LSTM, Dense, Dropout,
    MultiHeadAttention, LayerNormalization,
    GlobalAveragePooling1D, Add
)
from tensorflow.keras.callbacks import EarlyStopping

# 1. Reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# 2. Select existing aggregated dataset
data = df_60s.copy()

if "time_seconds" not in data.columns:
    data["time_seconds"] = (
        data.index.total_seconds()
        if isinstance(data.index, pd.TimedeltaIndex)
        else np.arange(len(data)) * 60
    )

features = [
    "total_cpu_demand",
    "mean_cpu_demand",
    "max_cpu_demand",
    "active_instances",
    "active_machines",
    "observation_count"
]

target = "total_cpu_demand"

# 3. Validate columns
missing = [c for c in features if c not in data.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

# 4. Clean and sort observations
data = data[["time_seconds"] + features].copy()

for col in ["time_seconds"] + features:
    data[col] = pd.to_numeric(data[col], errors="coerce")

data = data.replace([np.inf, -np.inf], np.nan)
data = data.sort_values("time_seconds")
data = data.dropna(subset=["time_seconds"])
data = data.drop_duplicates(subset=["time_seconds"], keep="last")

# Interpolate missing workload values
data[features] = data[features].interpolate(
    method="linear", limit_direction="both"
)

data = data.dropna(subset=features).reset_index(drop=True)

print("FINAL WORKLOAD DATASET")
print("Rows:", len(data))
print("Features:", features)
print(data[features].describe())

# 5. Chronological 80:20 train-test split
split = int(len(data) * 0.80)

if split <= 20 or len(data) - split <= 0:
    raise ValueError("Insufficient observations for sequence generation.")

# Fit scalers only on training observations
x_scaler = MinMaxScaler()
y_scaler = MinMaxScaler()

x_scaler.fit(data[features].iloc[:split])
y_scaler.fit(data[[target]].iloc[:split])

X_scaled = x_scaler.transform(data[features])
y_scaled = y_scaler.transform(data[[target]])

# 6. Generate 20-step sequences
WINDOW = 20

X, y, target_positions = [], [], []

for i in range(WINDOW, len(data)):
    X.append(X_scaled[i-WINDOW:i])
    y.append(y_scaled[i, 0])
    target_positions.append(i)

X = np.asarray(X, dtype=np.float32)
y = np.asarray(y, dtype=np.float32)
target_positions = np.asarray(target_positions)

train_mask = target_positions < split
test_mask = target_positions >= split

X_train, y_train = X[train_mask], y[train_mask]
X_test, y_test = X[test_mask], y[test_mask]

print("\nSEQUENCE GENERATION")
print("Input shape:", X.shape)
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# 7. Attention-Enhanced LSTM model
inputs = Input(shape=(WINDOW, len(features)))

lstm_output = LSTM(
    128, return_sequences=True
)(inputs)

attention_output = MultiHeadAttention(
    num_heads=4,
    key_dim=32
)(
    query=lstm_output,
    value=lstm_output
)

combined = Add()([lstm_output, attention_output])
combined = LayerNormalization()(combined)

pooled = GlobalAveragePooling1D()(combined)
hidden = Dense(64, activation="relu")(pooled)
hidden = Dropout(0.2)(hidden)

outputs = Dense(1)(hidden)

model = Model(inputs, outputs)

model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

model.summary()

# 8. Train model
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=7,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    validation_split=0.15,
    epochs=50,
    batch_size=64,
    callbacks=[early_stop],
    shuffle=False,
    verbose=1
)

# 9. Prediction
pred_scaled = model.predict(X_test, verbose=0)

y_actual = y_scaler.inverse_transform(
    y_test.reshape(-1, 1)
).ravel()

y_predicted = y_scaler.inverse_transform(
    pred_scaled
).ravel()

# 10. Evaluation
mae = mean_absolute_error(y_actual, y_predicted)
rmse = np.sqrt(mean_squared_error(y_actual, y_predicted))
r2 = r2_score(y_actual, y_predicted)

print("\nAELSTM MODEL PERFORMANCE")
print("=" * 45)
print("MAE :", round(mae, 6))
print("RMSE:", round(rmse, 6))
print("R2  :", round(r2, 6))

# 11. Save prediction results
os.makedirs("/content/AELSTM_Results", exist_ok=True)

results = pd.DataFrame({
    "time_seconds": data["time_seconds"].iloc[
        target_positions[test_mask]
    ].values,
    "actual_cpu_demand": y_actual,
    "predicted_cpu_demand": y_predicted
})

results.to_csv(
    "/content/AELSTM_Results/aelstm_predictions.csv",
    index=False
)

model.save(
    "/content/AELSTM_Results/aelstm_model.keras"
)

# 12. Actual vs predicted graph
plt.figure(figsize=(14, 5))
plt.plot(y_actual[:500], label="Actual CPU Demand")
plt.plot(y_predicted[:500], label="Predicted CPU Demand")
plt.xlabel("Test Observation")
plt.ylabel("CPU Demand")
plt.title("Attention-Enhanced LSTM: Actual vs Predicted Workload")
plt.legend()
plt.tight_layout()
plt.savefig(
    "/content/AELSTM_Results/actual_vs_predicted.png",
    dpi=300
)
plt.show()

# 13. Training loss graph
plt.figure(figsize=(10, 5))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("AELSTM Training and Validation Loss")
plt.legend()
plt.tight_layout()
plt.savefig(
    "/content/AELSTM_Results/training_loss.png",
    dpi=300
)
plt.show()

print("\nEXPERIMENT OUTPUTS SAVED")
print("/content/AELSTM_Results/")
print("1. aelstm_predictions.csv")
print("2. aelstm_model.keras")
print("3. actual_vs_predicted.png")
print("4. training_loss.png")
```

```python

from google.cloud import bigquery

query_sample = """
SELECT
    DIV(start_time, 300000000) AS time_bucket,
    COUNT(*) AS observation_count,
    SUM(average_usage.cpus) AS total_cpu_demand,
    AVG(average_usage.cpus) AS mean_cpu_demand
FROM
    `google.com:google-cluster-data.clusterdata_2019_a.instance_usage`
TABLESAMPLE SYSTEM (1 PERCENT)
WHERE
    start_time >= 0
    AND start_time < 86400000000
GROUP BY
    time_bucket
ORDER BY
    time_bucket
"""

# Dry run only: no dataset download
job_config = bigquery.QueryJobConfig(dry_run=True)

job = client.query(query_sample, job_config=job_config)

estimated_gb = job.total_bytes_processed / (1024**3)

print("1% SAMPLE DRY RUN")
print("-------------------------")
print("Estimated scan (GB):", round(estimated_gb, 3))
print("No data downloaded.")
```

```python

from google.cloud import bigquery
import os
import pandas as pd

# Google Drive project directory
BASE = "/content/drive/MyDrive/AELSTM_Cloud_Scheduling"
DATA_DIR = os.path.join(BASE, "Dataset")
os.makedirs(DATA_DIR, exist_ok=True)

# 10-second workload aggregation
query = """
SELECT
    DIV(start_time, 10000000) * 10 AS time_seconds,
    COUNT(*) AS observation_count,
    SUM(average_usage.cpus) AS total_cpu_demand,
    AVG(average_usage.cpus) AS mean_cpu_demand,
    MAX(average_usage.cpus) AS max_cpu_demand
FROM
    `google.com:google-cluster-data.clusterdata_2019_a.instance_usage`
TABLESAMPLE SYSTEM (1 PERCENT)
WHERE
    start_time >= 0
    AND start_time < 86400000000
GROUP BY
    time_seconds
ORDER BY
    time_seconds
"""

# Safety limit: maximum 2 GB billed
job_config = bigquery.QueryJobConfig(
    maximum_bytes_billed=2 * 1024**3
)

print("Starting Google Cluster Trace extraction...")

df_workload = client.query(
    query,
    job_config=job_config
).to_dataframe()

# Save immediately to Google Drive
output_file = os.path.join(
    DATA_DIR,
    "google_cluster_workload_1pct_24h.csv"
)

df_workload.to_csv(output_file, index=False)

print("\nEXTRACTION COMPLETED")
print("Dataset shape:", df_workload.shape)
print("Saved file:", output_file)
print("\nDataset preview:")
display(df_workload.head(10))

print("\nMissing values:")
print(df_workload.isnull().sum())
```
